Aplicamos `MinMaxScaler` al dataset `heart.csv`. La normalización lleva todos los atributos al rango [0, 1] para que ninguno domine por su escala.

```python
import numpy as np
import pandas as pd
from sklearn import preprocessing
```

```python
# Carga de datos
df = pd.read_csv('../dataset/heart.csv')
X = df.drop(columns=['target'])
print('Forma:', X.shape)
X.head()
```

```python
# Vemos los rangos originales (min, max) para confirmar la diferencia de escalas
print(X.agg(['min', 'max', 'mean']).round(2))
```

```python
# Limpieza de datos: normalización (MinMax)
normalizer = preprocessing.MinMaxScaler()
X_norm = normalizer.fit_transform(X)
X_norm = pd.DataFrame(X_norm, columns=X.columns)
X_norm.head()
```

```python
# Verificación: todas las columnas deben ir de 0 a 1
print(X_norm.agg(['min', 'max']).round(2))
```

```python
# Demostración de fit / transform por separado (igual que en clase)
f = normalizer.fit(X)
X_norm_2 = f.transform(X)
print('Iguales?', np.array_equiv(X_norm.values, X_norm_2))
```

### Conclusión
Todas las variables del dataset quedan en [0, 1]. La normalización es útil si vamos a usar algoritmos sensibles a la escala (KNN, regresión logística). Para nuestro modelo final usaremos **estandarización** (ver Programa 2), que suele dar mejores resultados con SVC.
