## Maching Learning - Estandarización (Heart Disease)

Aplicamos `StandardScaler` al dataset `heart.csv`. La estandarización transforma cada atributo para que tenga **media 0 y desviación estándar 1** (Z-score). Es la técnica que usaremos en el modelo final.

```python
import numpy as np
import pandas as pd
from sklearn import preprocessing
```

```python
# Carga de datos
df = pd.read_csv('../dataset/heart.csv')
X = df.drop(columns=['target'])
X.head()
```

```python
# Limpieza de datos: estandarización (Z-score)
standardizer = preprocessing.StandardScaler()
X_std = standardizer.fit_transform(X)
X_std = pd.DataFrame(X_std, columns=X.columns)
X_std.head()
```

```python
# Verificación: media ≈ 0 y desviación ≈ 1 en cada columna
print('Media:', X_std.mean().round(4).to_dict())
print('Std :', X_std.std().round(4).to_dict())
```

```python
# Demostración fit / transform por separado (como en clase)
f = preprocessing.StandardScaler()
f = f.fit(X)
X_std_2 = f.transform(X)
print('Iguales?', np.array_equal(X_std.values, X_std_2))
```

### Conclusión
Todas las variables tienen media 0 y desviación 1. Esta es la técnica que usaremos en el pipeline final porque el SVC con kernel RBF es muy sensible a la escala. El `scaler` entrenado aquí se guarda y se reutiliza en `app.py` para escalar los datos nuevos del formulario.
