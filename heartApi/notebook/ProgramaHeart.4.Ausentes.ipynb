Verificamos valores ausentes y aplicamos `SimpleImputer` (como en el Programa 4 original).
Adicionalmente revisamos duplicados, ya que en este dataset son un problema grave.

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
```

## 1) Valores ausentes

```python
df = pd.read_csv('../dataset/heart.csv')
print('Forma:', df.shape)
print('Nulos por columna:')
print(df.isnull().sum())
```

**El dataset no tiene valores nulos.** De todas formas, mostramos cómo se aplicaría el imputador por si llegara un CSV con ausentes.

```python
# Demostración: simulamos un dataset con algunos NaN y aplicamos imputación por la media
X = df.drop(columns=['target']).copy()
X_demo = X.iloc[:5].copy()
X_demo.iloc[0, 0] = np.nan  # ponemos un NaN en age
X_demo.iloc[2, 3] = np.nan  # ponemos un NaN en trestbps
print('Antes:'); print(X_demo)

imputer = SimpleImputer(strategy='mean')
X_demo_clean = pd.DataFrame(imputer.fit_transform(X_demo), columns=X_demo.columns)
print('\nDespues (media):'); print(X_demo_clean)
```

## 2) Duplicados (problema grave en este dataset)
El dataset de Kaggle `johnsmith88/heart-disease-dataset` contiene muchas filas idénticas. Si no las quitamos, un mismo paciente aparece en train **y** en test al mismo tiempo (data leakage), lo que infla artificialmente la exactitud del modelo.

```python
total = len(df)
dup = df.duplicated().sum()
print(f'Filas totales:     {total}')
print(f'Filas duplicadas:  {dup}  ({dup/total*100:.1f}%)')
print(f'Filas unicas:      {total - dup}')
```

```python
# Ejemplo concreto: paciente que mas se repite
conteo = df.value_counts().reset_index(name='veces')
print('Paciente mas repetido (aparece', conteo.iloc[0]['veces'], 'veces):')
print(conteo.iloc[0].drop('veces').to_dict())
```

```python
# Eliminamos duplicados y guardamos el dataset limpio
df_clean = df.drop_duplicates().reset_index(drop=True)
print(f'Pacientes unicos: {len(df_clean)}')
print(f'Distribucion target: {df_clean["target"].value_counts().to_dict()}')

df_clean.to_csv('../dataset/heart_clean.csv', index=False)
print('Dataset limpio guardado en dataset/heart_clean.csv')
```

### Conclusión
- **Valores ausentes:** ninguno; no se requiere imputación.
- **Duplicados:** 70.5% del dataset eran filas repetidas. Generamos `heart_clean.csv` con 302 pacientes únicos. Este será el dataset usado en los notebooks 5, 6 y 7.
