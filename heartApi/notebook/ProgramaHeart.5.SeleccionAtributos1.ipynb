## Selección de Atributos 1 - VarianceThreshold (Heart Disease)

Aplicamos `VarianceThreshold` al dataset limpio (`heart_clean.csv`). Esta técnica elimina atributos cuya varianza es menor a un umbral, asumiendo que aportan poca información.

```python
import numpy as np
import pandas as pd
from sklearn.feature_selection import VarianceThreshold
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv('../dataset/heart_clean.csv')
X = df.drop(columns=['target'])
cols = X.columns.tolist()
print('Forma:', X.shape)
X.head()
```

```python
# Varianza cruda y varianza sobre datos normalizados a [0,1]
# Es importante normalizar antes para comparar: si no, atributos con rangos grandes
# (chol va a 564) tendrian varianzas enormes que no reflejan su utilidad real.
X_mm = MinMaxScaler().fit_transform(X)
print(f"{'Atributo':<12} {'Var. cruda':>14} {'Var. norm[0-1]':>18}")
for i, c in enumerate(cols):
    print(f'{c:<12} {X[c].var():>14.4f} {X_mm[:,i].var():>18.4f}')
```

```python
# Aplicamos VarianceThreshold sobre datos normalizados con umbral th=0.02
th = 0.02
sel = VarianceThreshold(th)
sel.fit(X_mm)
mask = sel.get_support()
print(f'Umbral={th}')
print('Conservar:', [c for c,m in zip(cols, mask) if m])
print('Eliminar :', [c for c,m in zip(cols, mask) if not m])
```

### Conclusión
VarianceThreshold por sí solo marca como candidato a eliminar al atributo con menor varianza relativa. Es una primera criba; en el siguiente notebook usaremos técnicas más informativas (F-test y Información Mutua) que también consideran la relación atributo-target.
